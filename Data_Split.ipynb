# Train/Test Split

The dataset contains a field (strat_fold) that labels ten recommended folds. We will be using folds 1-8 for training, fold 9 for validation, and fold 10 for testing, as suggested in the data dictionary.

```python
from pyspark.sql import SparkSession
import os

spark = SparkSession.builder \
    .appName("Final_Project_Data_Split") \
    .getOrCreate()
```

```python
DATA_FILEPATH = '/standard/ds7200-apt4c/team2/ptb_superclass_v4'
```

```python
df = spark.read.parquet(DATA_FILEPATH)
df.select("ecg_id").show(5)
```

```python
df.groupby('strat_fold').count().orderBy('strat_fold').show()
```

```python
df.schema["strat_fold"].dataType
```

```python
train = df.filter("strat_fold <= 8")
validate = df.filter("strat_fold == 9")
test = df.filter("strat_fold == 10")
```
