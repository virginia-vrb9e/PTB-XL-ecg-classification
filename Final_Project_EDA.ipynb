# Exploratory Data Analysis

```python
from pyspark.sql import SparkSession
import os

spark = SparkSession.builder \
    .appName("Final_Project_EDA") \
    .getOrCreate()

DATA_FILEPATH = '/standard/ds7200-apt4c/team2/ptbxl_nw'
```

```python
df = spark.read.parquet(DATA_FILEPATH)
df.select("ecg_id").show(5)
```

```python
df.columns[1:23] #these are some of the non-technical columns
```

### Data Types

```python
from collections import Counter

dtype_counts = Counter(dtype for col, dtype in df.dtypes)

print(dtype_counts)
```

### Descriptive Statistics

```python
from pyspark.sql import functions as F

df_agecounts = (df
    .select("age")
    .groupby("age")
    .agg(F.count("age"))
    .orderBy("age")
               )
df_agecounts.show(100)
```
